```python
import os
import pandas as pd
import numpy as np

# Automatically find any CSV file under /kaggle/input
file_path = None
for root, dirs, files in os.walk('/kaggle/input'):
    for file in files:
        if file.endswith('.csv'):
            file_path = os.path.join(root, file)
            break
    if file_path:
        break

if file_path is None:
    raise FileNotFoundError("No CSV file found under /kaggle/input. Ensure the dataset is added under Input on the right sidebar.")

print(f"Loading file from: {file_path}")

# Load the customer data into a Pandas DataFrame
df = pd.read_csv(file_path)

# Data Cleaning Phase: Convert TotalCharges to numbers and drop empty slots
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df.dropna(subset=['TotalCharges'], inplace=True)

# Performance Evaluation Matrix: Group by Contract type to calculate Churn Ratio
churn_summary = df.groupby('Contract').agg(
    Total_Customers=('customerID', 'count'),
    Churned_Count=('Churn', lambda x: (x == 'Yes').sum())
).reset_index()

# Calculate the precise churn percentage rate
churn_summary['Churn_Rate_%'] = round((churn_summary['Churned_Count'] / churn_summary['Total_Customers']) * 100, 2)

print("\nPython Data Cleaning and Modeling Successful!\n")
print(churn_summary)
```
